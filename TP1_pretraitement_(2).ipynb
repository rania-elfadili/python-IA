```python
import re
import string

import nltk
nltk.download('punkt')
nltk.download('stopwords')

# on definie cette fonction pour découper le texte en mots selon la méthode choisit

def word_tokenizer(text, tokenizer_type):

    # la méthode nltk
    if tokenizer_type == 'nltk':
        from nltk.tokenize import word_tokenize
        return word_tokenize(text)
    # méthode simple avec spacy
    elif tokenizer_type == 'spacy':
        import spacy
        nlp = spacy.load("en_core_web_sm")
        doc = nlp(text)
        return [token.text for token in doc]
    # méthode split
    elif tokenizer_type == 'split':
        return text.split()

    else:
        return text.split()


# comme la fonction précédente mais il decoupe le texte en phrases selon la méthode choisit
def sentence_tokenizer(text, tokenizer_type):

    if tokenizer_type == 'nltk':
        from nltk.tokenize import sent_tokenize
        return sent_tokenize(text)

    elif tokenizer_type == 'spacy':
        import spacy
        nlp = spacy.load("en_core_web_sm")
        doc = nlp(text)
        return [sent.text for sent in doc.sents]

    elif tokenizer_type == 'split':
        return text.split(".")

    else:
        return text.split(".")

# on definie la fct normalize pour nettoyer et normaliser le texte (minuscule, ponctuation, espaces)

def normalize(text, operations):

    if "lowercase" in operations:
        text = text.lower()

    if "remove_punctuation" in operations:
        text = text.translate(str.maketrans('', '', string.punctuation))

    if "remove_extra_spaces" in operations:
        text = re.sub(r'\s+', ' ', text).strip()

    return text


#  cette fct permet de supprimer les mots vides inutiles

def remove_stop_words(tokens):

    from nltk.corpus import stopwords
    stop_words = set(stopwords.words('english'))

    return [word for word in tokens if word.lower() not in stop_words]


# on definie la fonction POS_tag pour identifier la catégorie grammaticale de chaque mot  (nom, verbe, etc)
def POS_tag(tokens):

    import spacy
    nlp = spacy.load("en_core_web_sm")

    doc = nlp(" ".join(tokens))
    return [token.pos_ for token in doc]


# cette fonction principale qui applique toutes les étapes de preprocessing

def process_text(text, sentence_tokenizer_type, word_tokenizer_type, operations, use_stopwords, use_pos,use_stemming, use_lemmatization):

    # initialisation des outils
    from nltk.stem import SnowballStemmer
    import spacy

    stemmer = SnowballStemmer("english")
    nlp = spacy.load("en_core_web_sm")

    # découper en phrases
    sentences = sentence_tokenizer(text, sentence_tokenizer_type)

    for i, sentence in enumerate(sentences, 1):

        print("\n======================")
        print("Phrase", i)

        # normalisation du texte
        normalized = normalize(str(sentence), operations)
        print("Texte normalisé :", normalized)

        # découper en mots
        tokens = word_tokenizer(normalized, word_tokenizer_type)

        # supprimer stopwords si activé
        if use_stopwords:
            tokens = remove_stop_words(tokens)

        # POS tagging si activé
        pos_tags = POS_tag(tokens) if use_pos else [""] * len(tokens)

        # lemmatization si activé
        doc = nlp(" ".join(tokens)) if use_lemmatization else None

        print("\nTableau :")

        for j, token in enumerate(tokens):

            row = {"Token": token}

            if use_pos:
                row["POS"] = pos_tags[j]

            if use_stemming:
                row["Stem"] = stemmer.stem(token)

            if use_lemmatization:
                row["Lemma"] = doc[j].lemma_

            print(row)


# maintenant on teste le programme avec des exemples
if __name__ == "__main__":

    text1 = "Hello! I am Rania El Fadili , from morocco ."
    text2 = "TP 1 prétraitement ."

    operations = ["lowercase", "remove_punctuation", "remove_extra_spaces"]

    print("\n===== TEST 1 =====")
    process_text(
        text1,
        sentence_tokenizer_type="nltk",
        word_tokenizer_type="nltk",
        operations=operations,
        use_stopwords=True,
        use_pos=True,
        use_stemming=True,
        use_lemmatization=True
    )

    print("\n===== TEST 2 =====")
    process_text(
        text2,
        sentence_tokenizer_type="spacy",
        word_tokenizer_type="spacy",
        operations=operations,
        use_stopwords=True,
        use_pos=True,
        use_stemming=True,
        use_lemmatization=True
    )
```
